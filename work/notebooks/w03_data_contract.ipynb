## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

This dataset does not include every possible SEO factor.

The sample loaded into memory is only a small subset of the complete warehouse.

External ranking factors such as backlinks, competitors, and Google algorithm updates are not included.

The final month should not be used for feature engineering because it represents the outcome period and may introduce data leakage.

```python
# Five Feature Frame

feature_df = df[[
    'gsc_impressions',
    'gsc_clicks',
    'gsc_avg_position',
    'ga4_pageviews',
    'ga4_sessions'
]]

feature_df.head()
```

Five Features Used

1. gsc_impressions
Available because impressions are already known when making the prediction.

2. gsc_clicks
Available because historical click data exists before prediction.

3. gsc_avg_position
Available because average search position is recorded before the prediction date.

4. ga4_pageviews
Available because pageviews are historical analytics data.

5. ga4_sessions
Available because session counts are available before making any prediction.

```python
# Deliberate Leakage Example

feature_df["leak_feature"] = df["sessions_organic"]

feature_df.head()
```

```python
# Remove leakage

feature_df = feature_df.drop(columns=["leak_feature"])

feature_df.head()
```

Leakage Experiment

I intentionally added the target variable (sessions_organic) as an input feature.

This would produce unrealistically high model performance because the model already knows the answer.

This feature must be removed before model training.

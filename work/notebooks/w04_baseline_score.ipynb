## 1. My rule and its reason codes

## My Baseline Rule

My baseline rule identifies content that is a good candidate for refresh.

The rule prioritizes pages that have:
- High search impressions (strong visibility)
- Low click-through rate (CTR), indicating users are not clicking despite seeing the page
- Low average position, suggesting ranking improvements could increase traffic

The purpose of this rule is to create a ranked list of content that should be reviewed for optimization before applying a machine learning model.

### Reason Codes

- HIGH_IMPRESSIONS_LOW_CTR
- LOW_RANKING_HIGH_IMPRESSIONS
- REFRESH_RECOMMENDED

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
!git clone https://github.com/Abhaykr45/flyrank-ml-internship.git
```

```python
import pandas as pd

df = pd.read_csv("flyrank-ml-internship/data/raw/content_refresh_anonymized.csv")
```

```python
print(df.columns)
```

```python
print(df.columns.tolist())
```

```python
# Baseline Score

df["score"] = (
    (df["impressions_90d"] * 0.5)
    + ((100 - df["ctr"]) * 0.3)
    + (df["avg_position"] * 0.2)
)

df[["impressions_90d", "ctr", "avg_position", "score"]].head()
```

```python
# Reason Code

df["reason_code"] = "REFRESH_RECOMMENDED"

# Action Label

df["action"] = "Refresh Content"

df[["score", "reason_code", "action"]].head()
```

```python
# Rank by score

df = df.sort_values("score", ascending=False)

df.head(10)
```

```python
import os

os.makedirs("work/outputs", exist_ok=True)

df.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

print("CSV created successfully!")
```
